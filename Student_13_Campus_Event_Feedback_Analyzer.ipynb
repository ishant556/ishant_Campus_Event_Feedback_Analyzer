## 4. NumPy practice
Convert one or more numeric columns to a NumPy array. Try `np.mean()`, `np.max()` and `np.min()` where appropriate.


```python
# TODO: Select numeric columns relevant to your project
# TODO: Convert them to a NumPy array using .to_numpy()
# TODO: Calculate average, maximum and minimum
rating_cols = ["Content_Rating", "Venue_Rating", "Organization_Rating"]

ratings_array = df[rating_cols].to_numpy()
print("Overall Mean Rating:", np.mean(ratings_array))
print("Overall Max Rating:", np.max(ratings_array))
print("Overall Min Rating:", np.min(ratings_array))
col_means = np.mean(ratings_array, axis=0)
for col, avg in zip(rating_cols, col_means):
    print(f"Average {col} (NumPy): {avg:.2f}")
```

## 5. Pandas analysis
Create at least two calculated columns or useful summaries. Filter and sort rows to answer questions about your dataset.


```python
# TODO: Create calculated column(s)
# TODO: Filter records using a condition
# TODO: Sort records and display the top/bottom results
df["Average_Rating"] = df[["Content_Rating", "Venue_Rating", "Organization_Rating"]].mean(axis=1)
df["Total_Rating"] = df["Content_Rating"] + df["Venue_Rating"] + df["Organization_Rating"]

high_rated_df = df[df["Average_Rating"] >= 4.0]
print("Responses with Average_Rating >= 4.0:", len(high_rated_df))
display(high_rated_df.head())

sorted_df = df.sort_values(by="Average_Rating", ascending=False)
print("Top-performing feedback record:")
display(sorted_df.head(1))

print("Lowest-performing feedback record:")
display(sorted_df.tail(1))
```

## 6. Your individual project challenge
**Challenge:** Calculate average ratings; compare events; identify lowest-rated feedback area.

Write your solution below and explain what the result means.


```python
# TODO: Solve your assigned challenge
if "Average_Rating" not in df.columns:
    df["Average_Rating"] = df[["Content_Rating", "Venue_Rating", "Organization_Rating"]].mean(axis=1)

area_averages = df[["Content_Rating", "Venue_Rating", "Organization_Rating"]].mean().sort_values()
print("Average Rating by Feedback Area (lowest to highest):")
print(area_averages.round(2))

lowest_area = area_averages.index[0]
lowest_score = area_averages.iloc[0]
print(f"\nLowest-rated feedback area: {lowest_area} (Average: {lowest_score:.2f})")
print("-" * 40)

event_comparison = df.groupby("Event")[
    ["Content_Rating", "Venue_Rating", "Organization_Rating", "Average_Rating"]
].mean().round(2)

print("Event-wise Average Ratings Comparison:")
display(event_comparison)

best_event = event_comparison["Average_Rating"].idxmax()
worst_event = event_comparison["Average_Rating"].idxmin()
print(f"Highest-rated event overall: {best_event} ({event_comparison.loc[best_event, 'Average_Rating']:.2f})")
print(f"Lowest-rated event overall: {worst_event} ({event_comparison.loc[worst_event, 'Average_Rating']:.2f})")
```
